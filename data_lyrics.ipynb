# Предобрабатываем данные — достаем все метаданные, которые можем

```python
from ast import literal_eval
```

```python
sorted(df_base["group type"].dropna().unique())
```

```python
all_values = set()

for value in df_base["genres"].dropna():
    try:
        parsed = literal_eval(str(value))

        if isinstance(parsed, list):
            all_values.update(parsed)
        else:
            all_values.add(str(parsed))

    except Exception:
        all_values.add(str(value))

print(f"Уникальных значений: {len(all_values)}")

for v in sorted(all_values):
    print(v)
```

```python
def parse_artist_type(row):
    group_type = str(row["group type"]).strip().lower()

    try:
        genres = literal_eval(str(row["genres"]))
        if not isinstance(genres, list):
            genres = []
        genres = [g.lower() for g in genres]
    except:
        genres = []

    gender_type = None
    is_group = None

    # Пол

    if group_type in ["female group", "female soloist"]:
        gender_type = 1

    elif group_type in ["male group", "male soloist"]:
        gender_type = 0

    elif group_type == "coed group":
        gender_type = 3

    else:
        if "k-pop girl group" in genres:
            gender_type = 1

        elif (
            "k-pop boy group" in genres
            or "j-pop boy group" in genres
        ):
            gender_type = 0

    # Соло/группа

    if group_type in ["female soloist", "male soloist"]:
        is_group = 0

    elif group_type in [
        "female group",
        "male group",
        "coed group"
    ]:
        is_group = 1

    elif (
        "k-pop girl group" in genres
        or "k-pop boy group" in genres
        or "j-pop boy group" in genres
    ):
        is_group = 1

    return pd.Series([gender_type, is_group])
```

```python
df_base[["gender_type", "is_group"]] = df_base.apply(parse_artist_type, axis=1)
```

```python
print(df_base["gender_type"].value_counts(dropna=False))
print(df_base["is_group"].value_counts(dropna=False))
```
